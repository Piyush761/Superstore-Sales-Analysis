```python
import matplotlib.pyplot as plt
```

```python
yearly = df.groupby(df["Order Date"].dt.year)[["Sales", "Profit"]].sum()
```

```python
yearly.plot(kind="line", marker="o")
plt.title("Year-wise Sales and Profit")
plt.xlabel("Year")
plt.ylabel("Amount ($)")
plt.grid(True)
plt.show()
```

```python
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values()
```

```python
subcategory_profit.plot(kind="bar", figsize=(12, 6))
plt.title("Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Profit ($)")
plt.axhline(0)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Project ke liye final insight
The analysis identifies Tables as the largest loss-making sub-category, while Copiers generates the highest total profit. 
Tables' discount patterns should be investigated as a potential contributor to losses. 
The business should review pricing and discount policies for loss-making products while examining what makes high-profit sub-categories successful.
